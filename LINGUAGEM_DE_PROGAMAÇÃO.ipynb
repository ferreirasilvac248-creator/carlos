```python
def e_par(numero):
  if numero * 2 == 0:
    return True
  else:
    return False
```

```python
def e_primo(numero):
  if numero < 2:
    return False
  for i in range(2, int(numero**0.5) + 1):
    if numero % i == 0:
      return False
  return True
```

```python
numero  = 72468
teste = e_primo (numero)
```
